```python
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None
```

```python
# Creating nodes
root = Node(1)
root.left = Node(2)
root.right = Node(3)
root.left.left = Node(4)
root.left.right = Node(5)
root.right.right = Node(6)
```

### DFS

```python
def inorder(node):
    if node:
        inorder(node.left)
        print(node.value, end=' ')
        inorder(node.right)

inorder(root)
```

```python
def preorder(node):
    if node:
        print(node.value, end=' ')
        preorder(node.left)        
        preorder(node.right)

preorder(root)
```

```python
def postorder(node):
    if node:
        postorder(node.left)        
        postorder(node.right)
        print(node.value, end=' ')

postorder(root)
```

### BFS

```python
from collections import deque
```

```python
def bfs(root):
    if not root:
        return

    queue = deque([root])

    while queue:
        node = queue.popleft()
        print(node.value, end=' ')

        if node.left:
            queue.append(node.left)
        if node.right:
            queue.append(node.right)
```

```python
bfs(root)
```
