```python
def fibo():

    n0 = 0
    n1 = 1

    yield n0
    yield n1

    while True:
        next = n0 + n1
        n0 = n1
        n1 = next
        yield next
```

```python
for i in fibo():
    print(i)

    if i > 1000000:
        break
```

```python
def infinite_sequence():
    num = 0
    while True:
        yield num
        num += 1
```

```python
gen = infinite_sequence()
```

```python
for i in range(15):
    print(next(gen))
```

```python
### csv file reading
import csv

def line_by_line_reader(file_path):
    with open(file_path, 'r') as file:
        csv_reader = csv.DictReader(file)
        for row in csv_reader:
            yield row
```
