```python
from abc import ABC, abstractmethod

# =============================================================================
# 1. METACLASS (Metasinif)
# =============================================================================

"""
METACLASS NEDİR?
- Metaclass, siniflarin sinifidir
- Siniflar nasil nesneler yaratiyorsa, metaclass'lar da siniflar yaratir
- Python'da her sinif aslinda type metaclass'inin bir örneğidir
- Metaclass'lar sinif oluşturma sürecini kontrol etmemizi sağlar
"""

# Basit metaclass örneği
class MyMetaClass(type):
    """Özel bir metaclass"""
    
    def __new__(cls, name, bases, attrs):
        # Sinif oluşturulmadan önce çalişir
        print(f"Metaclass: {name} sinifi oluşturuluyor...")
        
        # Tüm method isimlerini büyük harfle başlat
        new_attrs = {}
        for key, value in attrs.items():
            if callable(value) and not key.startswith('__'):
                new_attrs[key.upper()] = value
            else:
                new_attrs[key] = value
        
        return super().__new__(cls, name, bases, new_attrs)
    
    def __init__(cls, name, bases, attrs):
        # Sinif oluşturulduktan sonra çalişir
        print(f"Metaclass: {name} sinifi başlatildi")
        super().__init__(name, bases, attrs)

# Metaclass kullanan sinif
class MyClass(metaclass=MyMetaClass):
    def hello(self):
        return "Merhaba!"
    
    def goodbye(self):
        return "Hoşçakal!"

# Singleton Pattern için Metaclass
class SingletonMeta(type):
    """Singleton pattern için metaclass"""
    _instances = {}
    
    def __call__(cls, *args, **kwargs):
        if cls not in cls._instances:
            cls._instances[cls] = super().__call__(*args, **kwargs)
        return cls._instances[cls]

class DatabaseConnection(metaclass=SingletonMeta):
    def __init__(self):
        self.connection = "Database bağlantisi kuruldu"
```

```python
# =============================================================================
# 2. ABSTRACT METHOD (Soyut Metod)
# =============================================================================

"""
ABSTRACT METHOD NEDİR?
- Alt siniflarda mutlaka implement edilmesi gereken metodlar
- ABC (Abstract Base Class) modülü ile kullanilir
- Soyut siniflardan direkt nesne oluşturulamaz
- Interface benzeri bir yapi sağlar
"""

# Soyut sinif örneği
class Animal(ABC):
    """Soyut hayvan sinifi"""
    
    def __init__(self, name):
        self.name = name
    
    @abstractmethod
    def make_sound(self):
        """Her hayvan ses çikarmali - zorunlu metod"""
        pass
    
    @abstractmethod
    def move(self):
        """Her hayvan hareket etmeli - zorunlu metod"""
        pass
    
    # Concrete method (opsiyonel)
    def sleep(self):
        return f"{self.name} uyuyor"

# Soyut sinifi implement eden konkret siniflar
class Dog(Animal):
    def make_sound(self):
        return f"{self.name} havliyor: Hav hav!"
    
    def move(self):
        return f"{self.name} koşuyor"

class Cat(Animal):
    def make_sound(self):
        return f"{self.name} miyavliyor: Miyav!"
    
    def move(self):
        return f"{self.name} süzülüyor"

# Daha karmaşik soyut sinif örneği
class Shape(ABC):
    """Geometrik şekil soyut sinifi"""
    
    @abstractmethod
    def area(self):
        """Alan hesaplama - zorunlu"""
        pass
    
    @abstractmethod
    def perimeter(self):
        """Çevre hesaplama - zorunlu"""
        pass
    
    @abstractmethod
    def draw(self):
        """Çizim - zorunlu"""
        pass

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height
    
    def area(self):
        return self.width * self.height
    
    def perimeter(self):
        return 2 * (self.width + self.height)
    
    def draw(self):
        return f"Dikdörtgen çiziliyor: {self.width}x{self.height}"
```

```python
# =============================================================================
# 3. STATIC METHOD (Statik Metod)
# =============================================================================

"""
STATIC METHOD NEDİR?
- Sinifa ait olan ama self veya cls parametresi almayan metodlar
- Ne instance'a ne de sinifa bağimlidir
- Utility fonksiyonlar için kullanilir
- @staticmethod decorator'i ile tanimlanir
"""

class MathUtils:
    """Matematik yardimci fonksiyonlari"""
    
    @staticmethod
    def add(a, b):
        """İki sayiyi toplar"""
        return a + b
    
    @staticmethod
    def multiply(a, b):
        """İki sayiyi çarpar"""
        return a * b
    
    @staticmethod
    def is_even(number):
        """Sayinin çift olup olmadiğini kontrol eder"""
        return number % 2 == 0
    
    @staticmethod
    def factorial(n):
        """Faktöriyel hesaplar"""
        if n <= 1:
            return 1
        return n * MathUtils.factorial(n - 1)

class StringUtils:
    """String yardimci fonksiyonlari"""
    
    @staticmethod
    def reverse_string(s):
        """String'i tersine çevirir"""
        return s[::-1]
    
    @staticmethod
    def count_vowels(s):
        """Sesli harf sayisini döndürür"""
        vowels = "aeiouAEIOU"
        return sum(1 for char in s if char in vowels)
    
    @staticmethod
    def is_palindrome(s):
        """Palindrom kontrolü"""
        s = s.lower().replace(' ', '')
        return s == s[::-1]
```

```python
    print("\n" + "=" * 50)
    print("STATIC METHOD TESTİ")
    print("=" * 50)
    
    # Static method testi
    print(f"Toplama: {MathUtils.add(5, 3)}")
    print(f"Çarpma: {MathUtils.multiply(4, 7)}")
    print(f"8 çift mi? {MathUtils.is_even(8)}")
    print(f"5! = {MathUtils.factorial(5)}")
    
    print(f"String ters: {StringUtils.reverse_string('merhaba')}")
    print(f"Sesli harf sayisi: {StringUtils.count_vowels('merhaba')}")
    print(f"Palindrom mu? {StringUtils.is_palindrome('madam')}")
```

```python
# =============================================================================
# 4. CLASS METHOD (Sinif Metodu)
# =============================================================================

"""
CLASS METHOD NEDİR?
- Sinifa ait olan ve cls parametresi alan metodlar
- Instance değil, sinifla çalişir
- Alternative constructor'lar için sik kullanilir
- @classmethod decorator'i ile tanimlanir
"""

class Person:
    """Kişi sinifi - class method örnekleri"""
    
    # Class variable
    population = 0
    
    def __init__(self, name, age):
        self.name = name
        self.age = age
        Person.population += 1
    
    @classmethod
    def get_population(cls):
        """Toplam nüfusu döndürür"""
        return cls.population
    
    @classmethod
    def from_string(cls, person_str):
        """String'den Person oluşturur (alternative constructor)"""
        name, age = person_str.split('-')
        return cls(name, int(age))
    
    @classmethod
    def from_birth_year(cls, name, birth_year):
        """Doğum yilindan Person oluşturur"""
        current_year = 2024
        age = current_year - birth_year
        return cls(name, age)
    
    @classmethod
    def create_baby(cls, name):
        """Bebek oluşturur"""
        return cls(name, 0)
    
    def __str__(self):
        return f"{self.name} ({self.age} yaşinda)"

class BankAccount:
    """Banka hesabi - class method örnekleri"""
    
    # Class variables
    interest_rate = 0.05
    total_accounts = 0
    
    def __init__(self, account_holder, initial_balance=0):
        self.account_holder = account_holder
        self.balance = initial_balance
        self.account_number = BankAccount.total_accounts + 1
        BankAccount.total_accounts += 1
    
    @classmethod
    def set_interest_rate(cls, new_rate):
        """Faiz oranini ayarlar"""
        cls.interest_rate = new_rate
    
    @classmethod
    def get_total_accounts(cls):
        """Toplam hesap sayisini döndürür"""
        return cls.total_accounts
    
    @classmethod
    def create_joint_account(cls, holder1, holder2, initial_balance=0):
        """Ortak hesap oluşturur"""
        joint_holder = f"{holder1} & {holder2}"
        return cls(joint_holder, initial_balance)
    
    def deposit(self, amount):
        """Para yatirir"""
        self.balance += amount
        return self.balance
    
    def withdraw(self, amount):
        """Para çeker"""
        if amount <= self.balance:
            self.balance -= amount
            return self.balance
        else:
            return "Yetersiz bakiye!"
    
    def apply_interest(self):
        """Faiz uygular"""
        self.balance *= (1 + self.interest_rate)
        return self.balance
    
    def __str__(self):
        return f"Hesap #{self.account_number}: {self.account_holder} - Bakiye: {self.balance}"
```

```python
print("\n" + "=" * 50)
print("CLASS METHOD TESTİ")
print("=" * 50)

# Class method testi
person1 = Person("Ali", 25)
person2 = Person.from_string("Ayşe-30")
person3 = Person.from_birth_year("Mehmet", 1990)
baby = Person.create_baby("Bebek")

print(f"Kişi 1: {person1}")
print(f"Kişi 2: {person2}")
print(f"Kişi 3: {person3}")
print(f"Bebek: {baby}")
print(f"Toplam nüfus: {Person.get_population()}")

# Banka hesabi testi
account1 = BankAccount("Ahmet Yilmaz", 1000)
account2 = BankAccount.create_joint_account("Ali", "Veli", 2000)

print(f"Hesap 1: {account1}")
print(f"Hesap 2: {account2}")
print(f"Toplam hesap sayisi: {BankAccount.get_total_accounts()}")

BankAccount.set_interest_rate(0.08)
account1.apply_interest()
print(f"Faiz sonrasi hesap 1: {account1}")
```
