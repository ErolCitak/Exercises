### Step 1. Import the necessary libraries

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/drinks.csv). 

### Step 3. Assign it to a variable called drinks.

```python
drinks = pd.read_csv('drinks.csv')
drinks.head()
```

```python
drinks.shape
```

### Step 4. Which continent drinks more beer on average?

```python
avg_beer_consumption = drinks['beer_servings'].values.mean()
print("Avg Consumption: {:.3f}".format(avg_beer_consumption))

avg_drink_mask = drinks['beer_servings'] > avg_beer_consumption
drinks[avg_drink_mask][['country', 'beer_servings']]
```

### Step 5. For each continent print the statistics for wine consumption.

```python
drinks.groupby(by=drinks['continent'])[['wine_servings']].describe()
```

### Step 6. Print the mean alcohol consumption per continent for every column

```python
drinks.groupby(by=drinks['continent']).describe()
```

### Step 7. Print the median alcohol consumption per continent for every column

```python
alcohol_columns = ['beer_servings', 'spirit_servings', 'wine_servings', 'total_litres_of_pure_alcohol']
drinks.groupby('continent')[alcohol_columns].median()

#drinks.groupby('continent').median(numeric_only=True)
```

### Step 8. Print the mean, min and max values for spirit consumption.
#### This time output a DataFrame

```python
mean_df = drinks.groupby('continent')[['spirit_servings']].mean() # returns dataframe; two columns: continent und sprit
mean_series = mean_df['spirit_servings']

min_df = drinks.groupby('continent')[['spirit_servings']].min()
min_series = min_df['spirit_servings']

max_df = drinks.groupby('continent')[['spirit_servings']].max()
max_series = max_df['spirit_servings']

sprit_df = pd.DataFrame({'mean': mean_series,
            'min': min_series,
            'max': max_series})

sprit_df.head(3)
```

```python
spirit_df = drinks.groupby('continent')['spirit_servings'].agg(['mean', 'min', 'max'])

spirit_df.head(3)
```
