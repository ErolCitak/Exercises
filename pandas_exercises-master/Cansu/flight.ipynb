```python
import pandas as pd
from datetime import datetime

# Örnek veriler (DataFrame'ler)
flights = pd.DataFrame({
    "FlightID": [101, 102, 103, 104, 105],
    "CustomerID": [1, 2, 3, 4, 5],
    "FlightDate": ["2024-06-01", "2024-06-10", "2024-04-20", "2024-03-15", "2024-02-05"],
    "FlightCost": [1200, 800, 1050, 1500, 500],
})

# Günümüz tarihini alıyoruz
current_date = datetime.now()

# FlightDate'i datetime formatına dönüştürüyoruz
flights['FlightDate'] = pd.to_datetime(flights['FlightDate'])

# Uçuş tarihine olan farkı hesaplıyoruz
flights['DateDifference'] = abs(flights['FlightDate'] - current_date)

# Tarihi en yakın olan iki müşteri
closest_flights = flights.nsmallest(2, 'DateDifference')

# Sonuçları yazdırıyoruz
print("Uçuş tarihi en yakın olan iki müşteri:")
print(closest_flights[['CustomerID', 'FlightDate', 'DateDifference']])
```

```python
import pandas as pd
from datetime import datetime, timedelta

# Örnek veriler (güncellenmiş)
flights = pd.DataFrame({
    "FlightID": [101, 102, 103, 104, 105, 106, 107, 108],
    "CustomerID": [1, 2, 3, 4, 5, 1, 2, 3],
    "FlightDate": ["2024-06-01", "2024-06-10", "2024-04-20", "2024-03-15", "2024-02-05", "2024-07-01", "2024-07-15", "2024-05-01"],
    "FlightCost": [1200, 800, 1050, 1500, 500, 1300, 850, 1100],
    "CampaignID": [10, 12, 14, 11, 15, 13, 16, 17]
})

# Günümüz tarihi
current_date = datetime.now()

# Son bir yıl için tarih
one_year_ago = current_date - timedelta(days=365)

# Uçuş tarihini datetime formatına dönüştürüyoruz
flights['FlightDate'] = pd.to_datetime(flights['FlightDate'])

# Son bir yıl içindeki uçuşları filtreliyoruz
flights_last_year = flights[flights['FlightDate'] >= one_year_ago]

# 1000 Euro'nun üzerinde uçuş satın almış müşterileri buluyoruz
high_spender_customers = flights_last_year[flights_last_year['FlightCost'] > 1000]

# Müşteri başına toplam harcamayı hesaplıyoruz
total_spent = high_spender_customers.groupby('CustomerID')['FlightCost'].sum().reset_index()

# Sonuçları yazdırıyoruz
print("1000 Euro üzerinde uçuş yapan müşterilerin toplam harcamaları:")
print(total_spent)
```

```python
import pandas as pd
from datetime import datetime, timedelta

# Örnek veriler (güncellenmiş)
flights = pd.DataFrame({
    "FlightID": [101, 102, 103, 104, 105, 106, 107, 108],
    "CustomerID": [1, 2, 3, 4, 5, 1, 2, 3],
    "FlightDate": ["2024-06-01", "2024-06-10", "2024-04-20", "2024-03-15", "2024-02-05", "2024-07-01", "2024-07-15", "2024-05-01"],
    "FlightCost": [1200, 800, 1050, 1500, 500, 1300, 850, 1100],
    "CampaignID": [10, 12, 14, 11, 15, 13, 16, 17]
})

# Günümüz tarihi
current_date = datetime.now()

# Uçuş tarihini datetime formatına dönüştürüyoruz
flights['FlightDate'] = pd.to_datetime(flights['FlightDate'])

# 1000 Euro'nun üzerinde uçuş yapanları seçiyoruz
high_spender_customers = flights[flights['FlightCost'] > 1000]

# Henüz gerçekleşmemiş uçuşları (günümüzden sonraki uçuşları) filtreliyoruz
future_flights = high_spender_customers[high_spender_customers['FlightDate'] > current_date]

# En yakın iki uçuşu seçiyoruz
nearest_flights = future_flights.nsmallest(2, 'FlightDate')

# Sonuçları yazdırıyoruz
print("1000 Euro'nun üzerindeki uçuşlardan henüz gerçekleşmemiş en yakın iki uçuş:")
print(nearest_flights[['CustomerID', 'FlightID', 'FlightDate', 'FlightCost']])
```

```python
import pandas as pd
from datetime import datetime, timedelta

# Örnek veriler (güncellenmiş)
flights = pd.DataFrame({
    "FlightID": [101, 102, 103, 104, 105, 106, 107, 108],
    "CustomerID": [1, 2, 3, 4, 5, 1, 2, 3],
    "FlightDate": ["2024-06-01", "2024-06-10", "2024-04-20", "2024-03-15", "2024-02-05", "2024-07-01", "2024-07-15", "2024-05-01"],
    "FlightCost": [1200, 800, 1050, 1500, 500, 1300, 850, 1100],
    "CampaignID": [10, 12, 14, 11, 15, 13, 16, 17]
})

# Günümüz tarihi
current_date = datetime.now()

# Uçuş tarihini datetime formatına dönüştürüyoruz
flights['FlightDate'] = pd.to_datetime(flights['FlightDate'])

# Bir aydan az kalmış uçuşları bulmak için tarih farkını hesaplıyoruz
one_month_ahead = current_date + timedelta(days=30)

# Uçuş tarihi bir aydan az kalmış uçuşları seçiyoruz
flights_within_one_month = flights[(flights['FlightDate'] > current_date) & (flights['FlightDate'] <= one_month_ahead)]

# Sonuçları yazdırıyoruz
print("Bir aydan az kalmış uçuşları olan müşteriler:")
print(flights_within_one_month[['CustomerID', 'FlightID', 'FlightDate', 'FlightCost']])
```
