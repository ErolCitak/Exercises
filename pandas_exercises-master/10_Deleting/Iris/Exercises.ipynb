### Step 1. Import the necessary libraries

```python
import pandas as pd
import numpy as np
```

### Step 2. Import the dataset from this [address](https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data). 

### Step 3. Assign it to a variable called iris

```python
col_names = [
    "sepal_length",
    "sepal_width",
    "petal_length",
    "petal_width",
    "class"
]

iris = pd.read_csv("https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data", header=None, names=col_names)
iris.head()
```

### Step 4. Create columns for the dataset

```python
# 1. sepal_length (in cm)
# 2. sepal_width (in cm)
# 3. petal_length (in cm)
# 4. petal_width (in cm)
# 5. class
```

### Step 5.  Is there any missing value in the dataframe?

```python
iris.describe()
```

```python
iris.isnull().sum()
```

### Step 6.  Lets set the values of the rows 10 to 29 of the column 'petal_length' to NaN

```python
iris.loc[10:29, "petal_length"] = pd.NA
#iris.iloc[10:30,2:3] = np.nan

iris.head(32)
```

### Step 7. Good, now lets substitute the NaN values to 1.0

```python
iris.fillna(1.0, inplace=True)

iris.head(32)
```

### Step 8. Now let's delete the column class

```python
iris.drop(["class"], axis=1, inplace=True)

iris.head(3)
```

### Step 9.  Set the first 3 rows as NaN

```python
iris.iloc[:3, :] = pd.NA

iris.head()
```

### Step 10.  Delete the rows that have NaN

```python
iris.dropna(how="any", inplace=True)

iris.head()
```

### Step 11. Reset the index so it begins with 0 again

```python
iris.reset_index()
```
