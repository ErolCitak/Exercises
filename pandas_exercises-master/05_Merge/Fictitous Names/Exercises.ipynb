```python
### Merge v Join
```

```python
df = pd.DataFrame({"name": ["A", "B", "C", "D"],
                   "city": ["NY", "LA", "CH", "BO"],
                   "age": [25, 30, 22, 34]})

df2 = pd.DataFrame({"name": ["A", "B", "C", "E"],
                   "city": ["NY", "LA", "CH", "BO"],
                   "salary": [4872, 7716, 6096, 3000]})
```

```python
df
```

```python
df2
```

```python
## Merge
how_types = ["left", "right", "inner", "outer"]

print("*"*20)
print(df)
print("-"*10)
print(df2)
print("*"*20)

for how_type in how_types:
    print("--> {}".format(how_type))
    print(pd.merge(df, df2, how=how_type))
    print("-"*20)
```

```python
pd.merge(df, df2, on="name")
```

```python
pd.merge(df, df2, on="name", suffixes=["_u", "_v"])
```

```python
pd.merge(df, df2, on="name", suffixes=["_u", "_v"], how="left")
```

```python
## Join uses "Index" instead of common columns like in "Merge"!
## If you want to use desired columns for joining then
##    then we need to first create custome indices
how_types = ["left", "right", "inner", "outer"]

print("*"*20)
print(df)
print("-"*10)
print(df2)
print("*"*20)

for how_type in how_types:
    print("--> {}".format(how_type))
    print(df.join(df2, how=how_type))
    print("-"*20)
```
