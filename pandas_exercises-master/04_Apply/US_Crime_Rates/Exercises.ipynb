### Step 5. Convert the type of the column Year to datetime64

```python
year_column = crime['Year']
year_column_datetime = pd.to_datetime(crime['Year'].astype('int64'), format='%Y')

crime['Year_datetime'] = year_column_datetime

crime.head(3)
```
